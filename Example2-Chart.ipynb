# Create an Excel workbook containing y = x**2 data and a line chart.

```python
from argparse import ArgumentParser
from pathlib import Path

from openpyxl import Workbook
from openpyxl.chart import LineChart, Reference
from openpyxl.chart.axis import ChartLines


def create_quadratic_chart(output_path: str | Path) -> Path:
    """Create and save a workbook charting y = x**2 for -3 <= x <= 3."""
    output_path = Path(output_path)

    workbook = Workbook()
    worksheet = workbook.active
    worksheet.title = "Quadratic"
    worksheet.append(("x", "y = x²"))

    for x in range(-3, 4):
        worksheet.append((x, x**2))

    chart = LineChart()
    chart.title = "y = x²"
    chart.style = 13
    chart.x_axis.title = "x"
    chart.y_axis.title = "y"
    chart.y_axis.majorGridlines = ChartLines()
    chart.height = 8
    chart.width = 14

    y_values = Reference(
        worksheet,
        min_col=2,
        min_row=1,
        max_row=worksheet.max_row,
    )
    x_values = Reference(
        worksheet,
        min_col=1,
        min_row=2,
        max_row=worksheet.max_row,
    )
    chart.add_data(y_values, titles_from_data=True)
    chart.set_categories(x_values)
    chart.legend = None

    worksheet.add_chart(chart, "D2")
    worksheet.column_dimensions["A"].width = 12
    worksheet.column_dimensions["B"].width = 12

    output_path.parent.mkdir(parents=True, exist_ok=True)
    workbook.save(output_path)
    return output_path


def main(output_file_path) -> None:
    created_path = create_quadratic_chart(output_file_path)
    print(f"Created {created_path.resolve()}")

main('with_chart_out.xlsx')
```
