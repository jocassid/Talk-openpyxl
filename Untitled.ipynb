```python
from csv import QUOTE_NONNUMERIC

TEMP_NAME_COLUMN = 'name_temp'

states_df = pd.read_csv(
    'states.csv',
    header=1,
    names=['State', TEMP_NAME_COLUMN],
    usecols=[0, 1],
    # quoting=QUOTE_NONNUMERIC
)
states_df['Name'] = states_df[TEMP_NAME_COLUMN].str.slice(2, -1)
states_df = states_df.drop(TEMP_NAME_COLUMN, axis='columns')
states_df.head(5)
```

```python
merged_df = pd.merge(df, states_df, on='State', how='inner')
merged_df.head(5)
```
